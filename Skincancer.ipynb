```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
import numpy as np
import cv2

# Cargar modelo base preentrenado
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))

# Congelar capas del modelo base
for layer in base_model.layers:
    layer.trainable = False

# Añadir capas personalizadas para detección de cáncer de piel
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
output = Dense(1, activation='sigmoid')(x)

model = Model(inputs=base_model.input, outputs=output)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Simular entrenamiento (en un caso real se usarían imágenes etiquetadas)
x_train = np.random.rand(10, 224, 224, 3)  # 10 imágenes aleatorias
y_train = np.random.randint(0, 2, 10)  # Etiquetas aleatorias (0 = sano, 1 = sospechoso)
model.fit(x_train, y_train, epochs=5, batch_size=2)

# Convertir modelo a TensorFlow Lite
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

# Guardar modelo
with open('skin_cancer_model.tflite', 'wb') as f:
    f.write(tflite_model)

print("Modelo convertido y guardado como 'skin_cancer_model.tflite'")
```
