```python
# Instalar librerías necesarias
!pip install -q gtts moviepy Pillow

# 1. Crear imagen con texto usando PIL
from PIL import Image, ImageDraw, ImageFont
import os

def create_text_image(text, output_path="text_overlay.png", size=(1280, 720), font_size=40):
    img = Image.new("RGBA", size, (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    try:
        font = ImageFont.truetype("DejaVuSans-Bold.ttf", font_size)
    except:
        font = ImageFont.load_default()

    text_lines = text.split('\n')
    y = size[1] // 2 - (len(text_lines) * font_size) // 2
    for line in text_lines:
        # Use textbbox instead of textsize
        bbox = draw.textbbox((0, 0), line, font=font)
        w = bbox[2] - bbox[0] # Calculate width from bounding box
        h = bbox[3] - bbox[1] # Calculate height from bounding box
        x = (size[0] - w) // 2
        draw.text((x, y), line, font=font, fill="white")
        y += h + 10

    img.save(output_path)

# 2. Generar voz con gTTS
from gtts import gTTS

def text_to_speech(text, lang='es', output_file="temp_audio.mp3"):
    tts = gTTS(text=text, lang=lang)
    tts.save(output_file)
    return output_file

# 3. Crear video combinando imagen de fondo + texto + audio
from moviepy.editor import *

def create_video_with_pil_text(text, image_path="static_image.jpg", audio_file="temp_audio.mp3",
                                output_video="output_video.mp4", duration=8):
    create_text_image(text, "text_overlay.png")

    bg_clip = ImageClip(image_path, duration=duration).resize(height=720)
    txt_clip = ImageClip("text_overlay.png", duration=duration)

    final = CompositeVideoClip([bg_clip, txt_clip.set_position("center")])
    final = final.set_audio(AudioFileClip(audio_file))
    final.write_videofile(output_video, fps=24, codec="libx264")
    return output_video

# 4. Descargar imagen de fondo
import urllib.request

def download_image(image_url, image_path):
    if not os.path.exists(image_path):
        urllib.request.urlretrieve(image_url, image_path)

# 5. Generar y descargar video en Colab
from google.colab import files

def generate_and_download_video(text):
    image_url = "https://via.placeholder.com/1280x720.jpg?text=Sample+Image"
    image_path = "static_image.jpg"
    download_image(image_url, image_path)

    audio_file = text_to_speech(text)
    output_video = create_video_with_pil_text(text, image_path=image_path, audio_file=audio_file)
    files.download(output_video)

# 6. Ejecutar
input_text = "El cambio climático es un problema global\nque afecta a todos los países."
generate_and_download_video(input_text)
```
