```python
from PIL import Image, ImageDraw
import random

# Configuración
width, height = 2000, 2000
dpi = 300
background_color = (20, 20, 20)  # Fondo oscuro
num_profiles = 12  # Número de "perfiles"
output_file = "sombras_perfiles.png"

# Crear imagen RGBA
img = Image.new("RGBA", (width, height), background_color)
draw = ImageDraw.Draw(img)

def random_color(alpha=120):
    """Genera un color brillante con transparencia"""
    return (
        random.randint(150, 255),
        random.randint(150, 255),
        random.randint(150, 255),
        alpha
    )

def draw_profile(x, y, size_factor, angle):
    """Dibuja un perfil abstracto estilo cabeza con cuello"""
    head_radius = int(100 * size_factor)
    neck_height = int(60 * size_factor)

    # Cabeza
    head_bbox = [
        x - head_radius,
        y - head_radius,
        x + head_radius,
        y + head_radius
    ]

    # Mandíbula / cuello
    neck_points = [
        (x - head_radius // 2, y + head_radius),
        (x, y + head_radius + neck_height),
        (x + head_radius // 2, y + head_radius)
    ]

    # Dibujar en imagen temporal para rotar
    temp_img = Image.new("RGBA", img.size, (0, 0, 0, 0))
    temp_draw = ImageDraw.Draw(temp_img)
    color = random_color()

    temp_draw.ellipse(head_bbox, fill=color)
    temp_draw.polygon(neck_points, fill=color)

    # Rotar
    temp_img = temp_img.rotate(angle, expand=False, resample=Image.BILINEAR)

    # Pegar en imagen principal
    img.paste(temp_img, (0, 0), temp_img)

# Generar perfiles aleatorios
for _ in range(num_profiles):
    x = random.randint(400, 1600)
    y = random.randint(400, 1600)
    size_factor = random.uniform(0.8, 1.2)
    angle = random.randint(-45, 45)
    draw_profile(x, y, size_factor, angle)

# Guardar imagen
img.save(output_file, dpi=(dpi, dpi))

print(f"✅ Imagen guardada como: {output_file}")
```
