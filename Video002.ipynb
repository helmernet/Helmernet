```python
# Paso 5: Función para crear video con fondo, texto y audio (voz + música)
def create_full_video(text, video_path, music_path, output_video="video_final.mp4"):
    # Generar voz
    voice_file = text_to_speech(text)
    voice_clip = AudioFileClip(voice_file)

    # Cargar música y ajustar duración
    music_clip = AudioFileClip(music_path).volumex(0.2)
    music_clip = music_clip.subclip(0, voice_clip.duration)

    # Mezclar música y voz
    mixed_audio = CompositeAudioClip([music_clip, voice_clip])

    # Cargar video de fondo
    bg_clip = VideoFileClip(video_path).subclip(0, voice_clip.duration).resize(height=720)

    # Crear texto animado
    txt_clip = (TextClip(text, fontsize=50, color='white', size=bg_clip.size,
                         method='caption', bg_color='transparent') # Cambiado de None a 'transparent'
                .set_duration(voice_clip.duration)
                .set_position('center')
                .fadein(1)
                .fadeout(1))

    # Composición final
    final_clip = CompositeVideoClip([bg_clip, txt_clip])
    final_clip = final_clip.set_audio(mixed_audio)

    # Exportar
    final_clip.write_videofile(output_video, fps=24, codec='libx264')
    return output_video

# Paso 6: Subir archivos desde el equipo
print("⬆️ Por favor sube el VIDEO DE FONDO (.mp4)")
uploaded_video = files.upload()

print("⬆️ Ahora sube la MÚSICA DE FONDO (.mp3)")
uploaded_music = files.upload()

# Obtener nombres de archivos
video_path = list(uploaded_video.keys())[0]
music_path = list(uploaded_music.keys())[0]

# Paso 7: Texto de entrada y generar video
texto = "La inteligencia artificial está revolucionando nuestras vidas."

final_file = create_full_video(texto, video_path, music_path)

# Paso 8: Descargar video final
files.download(final_file)
```

```python
# Paso 5: Crear video con fondo, texto, voz y duración definida
def create_video_with_background(text, video_path, duration_seconds, output_video="video_final.mp4"):
    # Generar voz desde el texto
    voice_file = text_to_speech(text)
    voice_clip = AudioFileClip(voice_file)

    # Determinar duración final del video
    final_duration = min(duration_seconds, voice_clip.duration)

    # Cargar video de fondo y recortarlo a duración deseada
    bg_clip = VideoFileClip(video_path).subclip(0, final_duration).resize(height=720)

    # Crear texto sin fondo
    txt_clip = (TextClip(text, fontsize=50, color='white', size=bg_clip.size,
                         method='caption', bg_color='transparent') # Cambiado None a 'transparent'
                .set_duration(final_duration)
                .set_position('center')
                .fadein(1)
                .fadeout(1))

    # Recortar audio a duración final
    voice_clip = voice_clip.subclip(0, final_duration)

    # Combinar video + texto + voz
    final_clip = CompositeVideoClip([bg_clip, txt_clip])
    final_clip = final_clip.set_audio(voice_clip)

    # Exportar video
    final_clip.write_videofile(output_video, fps=24, codec='libx264')
    return output_video

# Paso 6: Subir video de fondo
print("⬆️ Por favor sube el VIDEO DE FONDO (.mp4)")
uploaded_video = files.upload()
video_path = list(uploaded_video.keys())[0]

# Paso 7: Ingresar texto y duración
texto = input("📝 Escribe el texto que deseas convertir en voz:\n")
duracion = int(input("⏱️ Especifica la duración deseada del video (en segundos):\n"))

# Paso 8: Crear y descargar el video final
video_generado = create_video_with_background(texto, video_path, duracion)
files.download(video_generado)
```
